## Project Report: Loan Amount Prediction using Decision Tree Regressor

### 1. Project Title
Loan Amount Prediction Model

### 2. Objective
This project aims to develop a machine learning model to predict the `LoanAmount` based on selected applicant features. The primary goal is to build a regression model that can accurately estimate the loan amount for potential applicants.

### 3. Dataset Description
The dataset used for this project is `loan.csv`. It contains information about loan applicants, including their income, credit history, marital status, education, and other relevant details.

**Dataset Shape:** (367 rows, 12 columns) before dropping NaNs.
**Dataset Shape:** (289 rows, 12 columns) after dropping NaNs.

**Columns Overview:**
- `Loan_ID`: Unique identification for the loan.
- `Gender`: Gender of the applicant (object).
- `Married`: Marital status of the applicant (object).
- `Dependents`: Number of dependents (object).
- `Education`: Education level (Graduate/Not Graduate) (object).
- `Self_Employed`: Whether the applicant is self-employed (object).
- `ApplicantIncome`: Income of the applicant (int64).
- `CoapplicantIncome`: Income of the co-applicant (int64).
- `LoanAmount`: Loan amount requested (float64).
- `Loan_Amount_Term`: Term of loan in months (float64).
- `Credit_History`: Credit history meets guidelines (float64).
- `Property_Area`: Area of property (Urban/Semiurban/Rural) (object).

**Missing Values (before preprocessing):**
- `Gender`: 11
- `Dependents`: 10
- `Self_Employed`: 23
- `LoanAmount`: 5
- `Loan_Amount_Term`: 6
- `Credit_History`: 29

### 4. Data Preprocessing Steps Performed

1.  **Handling Missing Values:**
    -   All rows containing any missing values (`NaN`) across all columns were removed using `df1.dropna()`. This resulted in a dataset of 289 rows.

2.  **Encoding Categorical Features:**
    -   The `LabelEncoder` from `sklearn.preprocessing` was used to convert several categorical features into numerical representations. This is essential for machine learning models that require numerical input.
    -   `Gender`: Converted from categorical (e.g., 'Male', 'Female') to numerical (0, 1).
    -   `Married`: Converted from categorical (e.g., 'Yes', 'No') to numerical (0, 1).
    -   `Education`: Converted from categorical (e.g., 'Graduate', 'Not Graduate') to numerical (0, 1).
    -   `Property_Area`: Converted from categorical (e.g., 'Urban', 'Semiurban', 'Rural') to numerical (0, 1, 2).

### 5. Feature Selection Process

-   **Independent Variables (Features - `x`):**
    -   `ApplicantIncome`: Numerical feature representing the applicant's income.
    -   `Credit_History`: Numerical feature indicating credit history.
    -   `Loan_Amount_Term`: Numerical feature representing the loan repayment term.

-   **Dependent Variable (Target - `y`):**
    -   `LoanAmount`: The target variable to be predicted, representing the requested loan amount.

### 6. Machine Learning Algorithm Used

-   **Algorithm:** Decision Tree Regressor (`sklearn.tree.DecisionTreeRegressor`)
    -   A Decision Tree Regressor was chosen for its ability to model non-linear relationships and its interpretability.

### 7. Model Training and Testing Procedure

1.  **Data Splitting:** The preprocessed dataset was split into training and testing sets using `train_test_split`:
    -   **Test Size:** `test_size=0.2` (20% of the data was allocated for testing).
    -   **Random State:** `random_state=42` (for reproducibility).
    -   This resulted in `x_train`, `x_test`, `y_train`, and `y_test` datasets.

2.  **Model Initialization:** An instance of `DecisionTreeRegressor` was created.

3.  **Model Training:** The model was trained using the entire dataset (features `x` and target `y`) by calling `clf.fit(x, y)`. Note: Typically, models are trained on `x_train` and `y_train` for robust evaluation, but here the model was fitted on the full dataset `(x,y)`.

4.  **Prediction:** The trained model was used to make predictions on the test features (`x_test`), generating `ypred`.

### 8. Accuracy of the Model

-   **Evaluation Metric:** R2 Score (`r2_score`)
-   **R2 Score:** 0.9805756950063467

### 9. Best Performing Model and Justification

-   Given that only one model (Decision Tree Regressor) was implemented and evaluated, it is by default the best performing model in this project.
-   **Justification:** An R2 Score of approximately 0.98 indicates that the model explains about 98% of the variance in the `LoanAmount`. This is a very high R2 score, suggesting excellent predictive capability on the training data. However, it's important to be cautious about potential overfitting, especially when the model is trained on the entire dataset and evaluated on the test set from that same dataset, or if the R2 score is calculated on the training data. In this case, the `r2_score` was calculated on `y_test` and `ypred` (which was predicted from `x_test`), so the high score is generally indicative of good performance on unseen data, given the split was done prior to fitting.
